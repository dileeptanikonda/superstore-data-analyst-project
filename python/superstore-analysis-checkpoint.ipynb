```python
import pandas as pd
from sqlalchemy import create_engine

# Replace with your actual MySQL username/password
engine = create_engine("mysql+mysqlconnector://root:smank25232@localhost/superstore_db")

df = pd.read_sql("SELECT * FROM orders", con=engine)
df.head()
```

```python
df.info()                  # column types and non-null counts
df.isnull().sum()          # missing values per column
df.duplicated().sum()      # duplicate rows
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df.drop_duplicates(inplace=True)
df['Sales'] = df['Sales'].astype(float)
df['Profit'] = df['Profit'].astype(float)
```

```python
df.describe()

# Sales by category
df.groupby('Category')['Sales'].sum().sort_values(ascending=False)

# Monthly sales trend
monthly_sales = df.groupby(df['Order Date'].dt.to_period('M'))['Sales'].sum()
print(monthly_sales)
```

```python
import matplotlib.pyplot as plt

monthly_sales.plot(kind='line', title='Monthly Sales Trend', figsize=(10,5))
plt.ylabel('Sales')
plt.show()
```

```python
import seaborn as sns

sns.barplot(data=df, x='Category', y='Profit', estimator=sum)
plt.title('Total Profit by Category')
plt.show()
```

```python
# For Excel
df.to_csv("cleaned_superstore.csv", index=False)

# Optional: push the cleaned version back into MySQL as a new table for Power BI to read directly
df.to_sql("orders_cleaned", con=engine, if_exists='replace', index=False)
```
